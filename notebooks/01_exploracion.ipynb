### Visualizacion de datos


```python
# Histograma de una variable para analizar la temperatura del aire

df["Air temperature [K]"].hist(bins=30)

plt.title("Distribucion de la temperatura del aire")
plt.xlabel("Temperatura (K)")
plt.ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
# Multiples histogramas para graficar las cinco variables de los sensores

variables = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
]

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for ax, columna in zip(axes, variables):
    ax.hist(df[columna], bins=30)
    ax.set_title(columna)
    ax.set_ylabel("Frecuencia")

for ax in axes[len(variables) :]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Boxplot para detectar valores extremos

fig, axes = plt.subplots(1, len(variables), figsize=(16, 5))

for ax, columna in zip(axes, variables):
    ax.boxplot(df[columna].dropna())
    ax.set_title(columna)
    ax.tick_params(axis="x", labelbottom=False)

plt.tight_layout()
plt.show()
```

```python
# Buscando correlaciones de Pearson entre variables

correlaciones = df[variables].corr()
display(correlaciones)
```

```python
# Grafico de las correlaciones

fig,ax = plt.subplots(figsize=(9,7))

imagen = ax.imshow(correlaciones, vmin=-1, vmax=1, cmap="coolwarm")

ax.set_xticks(range(len(variables)))
ax.set_xticklabels(variables, rotation=90)
ax.set_yticks(range(len(variables)))
ax.set_yticklabels(variables)

for i in range(len(variables)):
    for j in range(len(variables)):
        ax.text(
            j,i,
            f"{correlaciones.iloc[i,j]:.2f}",
            ha="center",
            va="center",
        )

fig.colorbar(imagen, ax=ax, label="Correlacion de Pearson")
ax.set_title("Correlacion entre variables de sensores")

plt.tight_layout()
plt.show()
```
